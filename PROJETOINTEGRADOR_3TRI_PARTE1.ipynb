```python
from IPython.display import HTML, display

html_code = """
<div id="cofre-app" style="font-family: Arial, sans-serif; background-color: #1e1e2e; color: #cdd6f4; padding: 20px; border-radius: 10px; max-width: 500px; margin: 0 auto;">
    <h3 style="color: #89b4fa; text-align: center; margin-top: 0;">🔐 Cofre Digital (JavaScript)</h3>

    <label style="font-size: 14px;"><b>1. Texto ou Código Criptografado:</b></label><br>
    <textarea id="txtEntrada" rows="3" style="width: 100%; box-sizing: border-box; margin: 6px 0 12px 0; background: #313244; color: #f5e0dc; border: 1px solid #45475a; border-radius: 6px; padding: 8px;" placeholder="Digite a mensagem ou cole o código aqui..."></textarea>

    <label style="font-size: 14px;"><b>2. Chave de Segurança:</b></label><br>
    <input type="password" id="txtChave" style="width: 100%; box-sizing: border-box; margin: 6px 0 16px 0; background: #313244; color: #f5e0dc; border: 1px solid #45475a; border-radius: 6px; padding: 8px;" placeholder="Digite a chave/senha">

    <div style="display: flex; gap: 10px; margin-bottom: 16px;">
        <button id="btnCripto" style="flex: 1; padding: 10px; background: #a6e3a1; color: #11111b; font-weight: bold; border: none; border-radius: 6px; cursor: pointer;">🔒 Criptografar</button>
        <button id="btnDescripto" style="flex: 1; padding: 10px; background: #f9e2af; color: #11111b; font-weight: bold; border: none; border-radius: 6px; cursor: pointer;">🔓 Descriptografar</button>
    </div>

    <label style="font-size: 14px;"><b>3. Resultado:</b></label><br>
    <textarea id="txtSaida" rows="3" readonly style="width: 100%; box-sizing: border-box; margin-top: 6px; background: #181825; color: #89b4fa; border: 1px solid #45475a; border-radius: 6px; padding: 8px; font-family: monospace;"></textarea>
</div>

<script>
(function() {
    function bytesToBase64(bytes) {
        let binString = Array.from(bytes, (byte) => String.fromCodePoint(byte)).join("");
        return btoa(binString);
    }

    function base64ToBytes(base64) {
        let binString = atob(base64);
        return Uint8Array.from(binString, (m) => m.codePointAt(0));
    }

    function criptografar() {
        const texto = document.getElementById('txtEntrada').value;
        const chave = document.getElementById('txtChave').value;
        const saida = document.getElementById('txtSaida');

        if (!texto || !chave) {
            saida.value = "Erro: Preencha o texto e a chave de segurança!";
            return;
        }

        try {
            const encoder = new TextEncoder();
            const bytesTexto = encoder.encode(texto);
            const bytesChave = encoder.encode(chave);

            let bytesCripto = new Uint8Array(bytesTexto.length);
            for (let i = 0; i < bytesTexto.length; i++) {
                bytesCripto[i] = (bytesTexto[i] + bytesChave[i % bytesChave.length]) % 256;
            }

            saida.value = bytesToBase64(bytesCripto);
        } catch (err) {
            saida.value = "Erro ao criptografar: " + err.message;
        }
    }

    function descriptografar() {
        const codigo = document.getElementById('txtEntrada').value.trim();
        const chave = document.getElementById('txtChave').value;
        const saida = document.getElementById('txtSaida');

        if (!codigo || !chave) {
            saida.value = "Erro: Cole o código e digite a chave de segurança!";
            return;
        }

        try {
            const bytesCripto = base64ToBytes(codigo);
            const encoder = new TextEncoder();
            const bytesChave = encoder.encode(chave);

            let bytesOriginais = new Uint8Array(bytesCripto.length);
            for (let i = 0; i < bytesCripto.length; i++) {
                bytesOriginais[i] = (bytesCripto[i] - bytesChave[i % bytesChave.length] + 256) % 256;
            }

            const decoder = new TextDecoder('utf-8', { fatal: true });
            saida.value = decoder.decode(bytesOriginais);
        } catch (err) {
            saida.value = "Erro: Código inválido ou chave incorreta!";
        }
    }

    document.getElementById('btnCripto').onclick = criptografar;
    document.getElementById('btnDescripto').onclick = descriptografar;
})();
</script>
"""

display(HTML(html_code))
```
